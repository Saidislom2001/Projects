```python
import shutil
import os

# Target the root tensorflow_datasets directory in Windows
tfds_root = os.path.expanduser('~/tensorflow_datasets')

found = False
if os.path.exists(tfds_root):
    for root, dirs, files in os.walk(tfds_root):
        if 'cats_vs_dogs' in dirs:
            target_path = os.path.join(root, 'cats_vs_dogs')
            print(f"Found corrupted folder at: {target_path}")
            shutil.rmtree(target_path)
            print("Successfully deleted!")
            found = True

if not found:
    print("No cats_vs_dogs folders found in ~/tensorflow_datasets.")
```
